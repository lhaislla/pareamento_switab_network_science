## Regras

```python
np.random.seed(42)

df['papel_aux'] = df['papel'].astype(str).str.strip().str.capitalize()
df['instituicao_aux'] = df['instituicao'].astype(str).str.strip().str.upper()
df['curso_aux'] = df['curso'].astype(str).str.strip().str.title()

mentoras = df[df['papel_aux'] == 'Mentora'].reset_index(drop=True)
mentoradas = df[df['papel_aux'] == 'Mentorada'].reset_index(drop=True)

print(f"--- Mapeamento de Participantes ---")
print(f"Total de Mentoras (Conjunto U): {len(mentoras)}")
print(f"Total de Mentoradas (Conjunto V): {len(mentoradas)}")
print(f"Coluna original 'papel' preservada intacta no dataframe principal.")
```

```python
def checa_mesma_instituicao(mentora, mentorada):
    """Regra 1: Mentora e Mentorada na mesma instituição de ensino"""
    return mentora['instituicao_aux'] == mentorada['instituicao_aux']

def checa_mesmo_curso(mentora, mentorada):
    """Regra 2: Mentora e Mentorada no mesmo curso de graduação"""
    return mentora['curso_aux'] == mentorada['curso_aux']

def checa_precedencia_periodo(mentora, mentorada):
    """Regra 3: Mentora em ano de ingresso anterior ou igual ao da mentorada"""
    try:
        return float(mentora['ano_ingresso']) <= float(mentorada['ano_ingresso'])
    except:
        return True

def calcula_compatibilidade_afinidade(mentora, mentorada):
    """Regra 4: Pontuação de Similaridade (Objetivos, Áreas, Soft Skills, Disponibilidade)"""
    def jaccard_str(str1, str2):
        if pd.isna(str1) or pd.isna(str2):
            return 0.0
        set1 = set([x.strip().lower() for x in str(str1).split(',')])
        set2 = set([x.strip().lower() for x in str(str2).split(',')])
        inter = set1.intersection(set2)
        uniao = set1.union(set2)
        return len(inter) / len(uniao) if len(uniao) > 0 else 0.0

    sim_carreira = jaccard_str(mentora.get('objetivo_carreira'), mentorada.get('objetivo_carreira'))
    sim_areas = jaccard_str(mentora.get('areas_interesse'), mentorada.get('areas_interesse'))
    sim_soft = jaccard_str(mentora.get('soft_skills'), mentorada.get('soft_skills'))
    sim_disp = jaccard_str(mentora.get('disponibilidade'), mentorada.get('disponibilidade'))

    # Média dos fatores de afinidade
    score = (0.35 * sim_carreira) + (0.35 * sim_areas) + (0.15 * sim_soft) + (0.15 * sim_disp)
    return score
```

## Rede Aleatória de Pareamento

### Baseline

**Objetivo:** Criar um modelo de pareamento aleatório sem restrições preliminares para servir de *baseline* (linha de base).
A partir dessa rede aleatória, calculamos:
1. **Taxa de Violação das Regras Rígidas:** Quantos pares aleatórios violam as regras institucionais (diferente campus, diferente curso, ou mentorada em período superior).
2. **Pontuação Média de Compatibilidade Aleatória:** Medida de afinidade (carreira, soft skills, disponibilidade) sob alocação ao acaso.
3. **Métricas Topológicas:** Densidade, Grau Médio e Coeficiente de Aglomeração Médio.

```python
def criar_rede_aleatoria_baseline(mentoras, mentoradas, **kwargs):
    """
    Constrói a rede aleatória (baseline) preservando os conjuntos U (mentoras) e V (mentoradas).
    """
    import numpy as np
    import networkx as nx

    G_aleatorio = nx.Graph()

    for _, row in mentoras.iterrows():
        G_aleatorio.add_node(row['id'], bipartite=0,
                             instituicao=row['instituicao'], curso=row['curso'], ano_ingresso=row['ano_ingresso'])

    for _, row in mentoradas.iterrows():
        G_aleatorio.add_node(row['id'], bipartite=1,
                             instituicao=row['instituicao'], curso=row['curso'], ano_ingresso=row['ano_ingresso'])

    violacoes_inst = 0
    violacoes_curso = 0
    violacoes_periodo = 0
    scores_compatibilidade = []

    mentoras_ids = mentoras['id'].tolist()
    total_pares = len(mentoradas)

    for _, mentorada in mentoradas.iterrows():
        mentora_id_sorteada = np.random.choice(mentoras_ids)
        mentora_sorteada = mentoras[mentoras['id'] == mentora_id_sorteada].iloc[0]

        G_aleatorio.add_edge(mentora_id_sorteada, mentorada['id'])

        if not checa_mesma_instituicao(mentora_sorteada, mentorada):
            violacoes_inst += 1
        if not checa_mesmo_curso(mentora_sorteada, mentorada):
            violacoes_curso += 1
        if not checa_precedencia_periodo(mentora_sorteada, mentorada):
            violacoes_periodo += 1

        score = calcula_compatibilidade_afinidade(mentora_sorteada, mentorada)
        scores_compatibilidade.append(score)

    media_compatibilidade = np.mean(scores_compatibilidade) if scores_compatibilidade else 0.0
    densidade = nx.density(G_aleatorio)
    clustering = nx.average_clustering(G_aleatorio)

    metricas = {
        'total_pares': total_pares,
        'total_arestas': G_aleatorio.number_of_edges(),
        'compatibilidade_media': media_compatibilidade,
        'violacoes_instituicao': violacoes_inst,
        'violacoes_curso': violacoes_curso,
        'violacoes_periodo': violacoes_periodo,
        'densidade': densidade,
        'clustering_medio': clustering
    }

    return G_aleatorio, metricas
```

```python
def exibir_metricas_baseline(metricas):
    """
    Recebe o dicionário de métricas gerado pela rede aleatória
    e exibe o relatório formatado no console.
    """
    total_pares = metricas['total_pares']
    v_inst = metricas['violacoes_instituicao']
    v_curso = metricas['violacoes_curso']
    v_periodo = metricas['violacoes_periodo']

    print("MÉTRICAS DA REDE ALEATÓRIA  ")
    print(f"• Total de Arestas (Duplas): {metricas['total_arestas']}")
    print(f"• Pontuação Média de Compatibilidade: {metricas['compatibilidade_media']:.4f}")
    print(f"• Taxa de Violações de Instituição: {v_inst} ({v_inst/total_pares*100:.1f}%)")
    print(f"• Taxa de Violações de Curso: {v_curso} ({v_curso/total_pares*100:.1f}%)")
    print(f"• Taxa de Violações de Período: {v_periodo} ({v_periodo/total_pares*100:.1f}%)")
    print(f"• Densidade da Rede: {metricas['densidade']:.4f}")
    print(f"• Coeficiente de Aglomeração Médio: {metricas['clustering_medio']:.4f}")
```

```python
def plotar_rede_dinamica(G):
    """
    Gera uma visualização interativa premium e estilizada da rede bipartida (SWITAB),
    com legenda lateral funcional e separação clara por grupos, blindada contra chaves ausentes.
    """
    import plotly.graph_objects as go
    import networkx as nx

    top_nodes = [n for n, d in G.nodes(data=True) if d.get('bipartite') == 0]
    pos = nx.bipartite_layout(G, nodes=top_nodes, align='horizontal')

    edge_x = []
    edge_y = []
    for edge in G.edges():
        x0, y0 = pos[edge[0]]
        x1, y1 = pos[edge[1]]
        edge_x.extend([x0, x1, None])
        edge_y.extend([y0, y1, None])

    edge_trace = go.Scatter(
        x=edge_x, y=edge_y,
        line=dict(width=0.6, color='#cbd5e1'),
        hoverinfo='none',
        mode='lines'
    )

    x_mentoras, y_mentoras, text_mentoras = [], [], []
    x_mentoradas, y_mentoradas, text_mentoradas = [], [], []

    for node in G.nodes():
        x, y = pos[node]
        bipartite_group = G.nodes[node].get('bipartite')

        # Busca robusta para o papel e atributos
        papel = str(
            G.nodes[node].get('papel') or
            G.nodes[node].get('pape_interesse') or
            'Desconhecido'
        ).strip().capitalize()

        inst = G.nodes[node].get('instituicao') or G.nodes[node].get('inst') or 'N/A'
        curso = G.nodes[node].get('curso', 'N/A')

        # Busca robusta para o ano de ingresso com múltiplos fallbacks
        ano = (
            G.nodes[node].get('ano_ingresso') or
            G.nodes[node].get('ano') or
            G.nodes[node].get('ano_de_ingresso') or
            'N/A'
        )

        tooltip = (
            f"<b>ID:</b> {node}<br>"
            f"<b>Papel na Base:</b> {papel}<br>"
            f"<b>Instituição:</b> {inst}<br>"
            f"<b>Curso:</b> {curso}<br>"
            f"<b>Ano Ingresso:</b> {ano}"
        )

        if bipartite_group == 0:
            x_mentoras.append(x)
            y_mentoras.append(y)
            text_mentoras.append(tooltip)
        else:
            x_mentoradas.append(x)
            y_mentoradas.append(y)
            text_mentoradas.append(tooltip)

    trace_mentoras = go.Scatter(
        x=x_mentoras, y=y_mentoras,
        mode='markers',
        name='Conjunto U (Mentoras / Veteranas)',
        hoverinfo='text',
        text=text_mentoras,
        marker=dict(
            color='#1d4ed8',
            size=12,
            symbol='circle',
            line=dict(width=1.5, color='#ffffff')
        )
    )

    trace_mentoradas = go.Scatter(
        x=x_mentoradas, y=y_mentoradas,
        mode='markers',
        name='Conjunto V (Mentoradas / Calouras)',
        hoverinfo='text',
        text=text_mentoradas,
        marker=dict(
            color='#e11d48',
            size=10,
            symbol='circle-open',
            line=dict(width=2, color='#e11d48')
        )
    )

    fig = go.Figure(
        data=[edge_trace, trace_mentoras, trace_mentoradas],
        layout=go.Layout(
            title=dict(
                text='<b>Rede Bipartida Interativa — SWITAB (Baseline)</b><br><sup>Clique na legenda para filtrar os grupos | Passe o mouse para ver os detalhes</sup>',
                font=dict(size=14, family='Inter, sans-serif', color='#1e293b'),
                x=0.05
            ),
            showlegend=True,
            legend=dict(
                orientation="h",
                yanchor="bottom",
                y=1.02,
                xanchor="right",
                x=1
            ),
            hovermode='closest',
            margin=dict(b=20, l=20, r=20, t=80),
            xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
            yaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
            plot_bgcolor='#ffffff',
            paper_bgcolor='#ffffff'
        )
    )

    fig.show()
```

### Rede aleatória por Institituição

```python
def executar_baseline_por_instituicao(df):
    """
    Executa a construção da rede aleatória (baseline) separadamente para
    cada instituição, garantindo que o atributo 'papel' seja injetado corretamente nos nós.
    """
    import numpy as np
    import networkx as nx

    mediana_ingresso = df['ano_ingresso'].median()
    mentoras_geral = df[df['ano_ingresso'] <= mediana_ingresso].reset_index(drop=True)
    mentoradas_geral = df[df['ano_ingresso'] > mediana_ingresso].reset_index(drop=True)

    instituicoes = df['instituicao'].dropna().unique()
    resultados_instituicoes = {}

    for inst in instituicoes:
        df_inst = df[df['instituicao'] == inst]
        mentoras_inst = mentoras_geral[mentoras_geral['instituicao'] == inst]
        mentoradas_inst = mentoradas_geral[mentoradas_geral['instituicao'] == inst]

        if len(mentoras_inst) == 0 or len(mentoradas_inst) == 0:
            continue

        G_inst = nx.Graph()

        for _, row in mentoras_inst.iterrows():
            G_inst.add_node(
                row['id'],
                bipartite=0,
                papel=row.get('papel', 'N/A'),
                instituicao=row['instituicao'],
                curso=row['curso'],
                ano_ingresso=row['ano_ingresso']
            )

        for _, row in mentoradas_inst.iterrows():
            G_inst.add_node(
                row['id'],
                bipartite=1,
                papel=row.get('papel', 'N/A'),
                instituicao=row['instituicao'],
                curso=row['curso'],
                ano_ingresso=row['ano_ingresso']
            )

        violacoes_periodo = 0
        scores_compatibilidade = []
        cursos = df_inst['curso'].dropna().unique()

        for curso in cursos:
            mentoras_curso = mentoras_inst[mentoras_inst['curso'] == curso]
            mentoradas_curso = mentoradas_inst[mentoradas_inst['curso'] == curso]

            if len(mentoras_curso) == 0 or len(mentoradas_curso) == 0:
                continue

            mentoras_ids_curso = mentoras_curso['id'].tolist()

            for _, mentorada in mentoradas_curso.iterrows():
                mentora_id_sorteada = np.random.choice(mentoras_ids_curso)
                mentora_sorteada = mentoras_curso[mentoras_curso['id'] == mentora_id_sorteada].iloc[0]

                G_inst.add_edge(mentora_id_sorteada, mentorada['id'])

                if float(mentora_sorteada['ano_ingresso']) > float(mentorada['ano_ingresso']):
                    violacoes_periodo += 1

                score = calcula_compatibilidade_afinidade(mentora_sorteada, mentorada)
                scores_compatibilidade.append(score)

        total_pares = len(G_inst.edges())
        media_compatibilidade = np.mean(scores_compatibilidade) if scores_compatibilidade else 0.0
        densidade = nx.density(G_inst) if G_inst.number_of_nodes() > 0 else 0.0
        clustering = nx.average_clustering(G_inst) if G_inst.number_of_nodes() > 0 else 0.0

        metricas = {
            'total_pares': total_pares,
            'total_arestas': total_pares,
            'compatibilidade_media': media_compatibilidade,
            'violacoes_instituicao': 0,
            'violacoes_curso': 0,
            'violacoes_periodo': violacoes_periodo,
            'densidade': densidade,
            'clustering_medio': clustering
        }

        resultados_instituicoes[inst] = {
            'grafo': G_inst,
            'metricas': metricas
        }

    return resultados_instituicoes
```

```python
def executar_baseline_por_instituicao(df):
    """
    Executa a construção da rede aleatória (baseline) separadamente para
    cada instituição, utilizando um corte cronológico absoluto por ano de
    ingresso para separar mentoras e mentoradas de forma equilibrada.
    """
    import numpy as np
    import networkx as nx

    df_proc = df.copy()
    df_proc['papel_dinamico'] = df_proc['pape_interesse'].astype(str).str.strip().str.capitalize()

    condicao_mentora_explícita = df_proc['papel_dinamico'] == 'Mentora'
    condicao_mentorada_explícita = df_proc['papel_dinamico'] == 'Mentorada'

    criterio_veterana_absoluto = df_proc['ano_ingresso'] < 2025

    df_proc['papel_final'] = 'Mentorada'
    df_proc.loc[condicao_mentora_explícita, 'papel_final'] = 'Mentora'
    df_proc.loc[df_proc['papel_dinamico'] == 'Primeira vez no projeto', 'papel_final'] = np.where(
        criterio_veterana_absoluto[df_proc['papel_dinamico'] == 'Primeira vez no projeto'],
        'Mentora',
        'Mentorada'
    )

    instituicoes = df_proc['instituicao'].dropna().unique()
    resultados_instituicoes = {}

    for inst in instituicoes:
        df_inst = df_proc[df_proc['instituicao'] == inst].copy()

        if len(df_inst) == 0:
            continue

        mentoras_inst = df_inst[df_inst['papel_final'] == 'Mentora'].reset_index(drop=True)
        mentoradas_inst = df_inst[df_inst['papel_final'] == 'Mentorada'].reset_index(drop=True)

        if len(mentoras_inst) == 0 or len(mentoradas_inst) == 0:
            continue

        G_inst = nx.Graph()

        for _, row in mentoras_inst.iterrows():
            G_inst.add_node(
                row['id'],
                bipartite=0,
                papel=row['papel_final'],
                instituicao=row['instituicao'],
                curso=row['curso'],
                ano_ingresso=row['ano_ingresso']
            )

        for _, row in mentoradas_inst.iterrows():
            G_inst.add_node(
                row['id'],
                bipartite=1,
                papel=row['papel_final'],
                instituicao=row['instituicao'],
                curso=row['curso'],
                ano_ingresso=row['ano_ingresso']
            )

        violacoes_periodo = 0
        scores_compatibilidade = []
        cursos = df_inst['curso'].dropna().unique()

        for curso in cursos:
            mentoras_curso = mentoras_inst[mentoras_inst['curso'] == curso]
            mentoradas_curso = mentoradas_inst[mentoradas_inst['curso'] == curso]

            if len(mentoras_curso) == 0 or len(mentoradas_curso) == 0:
                continue

            mentoras_ids_curso = mentoras_curso['id'].tolist()

            for _, mentorada in mentoradas_curso.iterrows():
                mentora_id_sorteada = np.random.choice(mentoras_ids_curso)
                mentora_sorteada = mentoras_curso[mentoras_curso['id'] == mentora_id_sorteada].iloc[0]

                G_inst.add_edge(mentora_id_sorteada, mentorada['id'])

                if float(mentora_sorteada['ano_ingresso']) > float(mentorada['ano_ingresso']):
                    violacoes_periodo += 1

                score = calcula_compatibilidade_afinidade(mentora_sorteada, mentorada)
                scores_compatibilidade.append(score)

        total_pares = len(G_inst.edges())
        media_compatibilidade = np.mean(scores_compatibilidade) if scores_compatibilidade else 0.0
        densidade = nx.density(G_inst) if G_inst.number_of_nodes() > 0 else 0.0
        clustering = nx.average_clustering(G_inst) if G_inst.number_of_nodes() > 0 else 0.0

        metricas = {
            'total_pares': total_pares,
            'total_arestas': total_pares,
            'compatibilidade_media': media_compatibilidade,
            'violacoes_instituicao': 0,
            'violacoes_curso': 0,
            'violacoes_periodo': violacoes_periodo,
            'densidade': densidade,
            'clustering_medio': clustering
        }

        resultados_instituicoes[inst] = {
            'grafo': G_inst,
            'metricas': metricas
        }

        print(f"INSTITUIÇÃO: {inst} ")
        print(f"• Total de Arestas (Duplas): {total_pares}")
        print(f"• Pontuação Média de Compatibilidade: {media_compatibilidade:.4f}")
        print(f"• Violações de Período: {violacoes_periodo}")
        print(f"• Densidade da Rede: {densidade:.4f}")
        print(f"• Coeficiente de Aglomeração Médio: {clustering:.4f}\n")

    return resultados_instituicoes
```

```python
def plotar_rede_por_instituicao(resultados_por_instituicao, instituicao_nome):
    """
    Recebe o dicionário de resultados e o nome da instituição desejada,
    gerando e exibindo a visualização interativa em Plotly para o campus específico,
    exibindo na legenda e nos tooltips o papel real registrado na base de dados.
    """
    import plotly.graph_objects as go
    import networkx as nx

    if instituicao_nome not in resultados_por_instituicao:
        print(f"Instituição '{instituicao_nome}' não encontrada nos resultados.")
        return

    G = resultados_por_instituicao[instituicao_nome]['grafo']

    if G.number_of_nodes() == 0:
        print(f"A rede da instituição '{instituicao_nome}' está vazia.")
        return

    top_nodes = [n for n, d in G.nodes(data=True) if d.get('bipartite') == 0]
    pos = nx.bipartite_layout(G, nodes=top_nodes, align='horizontal')

    edge_x = []
    edge_y = []
    for edge in G.edges():
        x0, y0 = pos[edge[0]]
        x1, y1 = pos[edge[1]]
        edge_x.extend([x0, x1, None])
        edge_y.extend([y0, y1, None])

    edge_trace = go.Scatter(
        x=edge_x, y=edge_y,
        line=dict(width=0.6, color='#cbd5e1'),
        hoverinfo='none',
        mode='lines'
    )

    x_mentoras, y_mentoras, text_mentoras = [], [], []
    x_mentoradas, y_mentoradas, text_mentoradas = [], [], []

    for node in G.nodes():
        x, y = pos[node]
        bipartite_group = G.nodes[node].get('bipartite')

        papel = str(G.nodes[node].get('papel', 'Desconhecido')).strip().capitalize()
        inst = G.nodes[node].get('instituicao', 'N/A')
        curso = G.nodes[node].get('curso', 'N/A')
        ano = G.nodes[node].get('ano_ingresso', 'N/A')

        tooltip = (
            f"<b>ID:</b> {node}<br>"
            f"<b>Papel na Base:</b> {papel}<br>"
            f"<b>Instituição:</b> {inst}<br>"
            f"<b>Curso:</b> {curso}<br>"
            f"<b>Ano Ingresso:</b> {ano}"
        )

        if bipartite_group == 0:
            x_mentoras.append(x)
            y_mentoras.append(y)
            text_mentoras.append(tooltip)
        else:
            x_mentoradas.append(x)
            y_mentoradas.append(y)
            text_mentoradas.append(tooltip)

    trace_mentoras = go.Scatter(
        x=x_mentoras, y=y_mentoras,
        mode='markers',
        name='Papel na Base: Mentora',
        hoverinfo='text',
        text=text_mentoras,
        marker=dict(
            color='#1d4ed8',
            size=12,
            symbol='circle',
            line=dict(width=1.5, color='#ffffff')
        )
    )

    trace_mentoradas = go.Scatter(
        x=x_mentoradas, y=y_mentoradas,
        mode='markers',
        name='Papel na Base: Mentorada',
        hoverinfo='text',
        text=text_mentoradas,
        marker=dict(
            color='#e11d48',
            size=10,
            symbol='circle-open',
            line=dict(width=2, color='#e11d48')
        )
    )

    fig = go.Figure(
        data=[edge_trace, trace_mentoras, trace_mentoradas],
        layout=go.Layout(
            title=dict(
                text=f'<b>Rede Bipartida — {instituicao_nome}</b><br><sup>Validação com o Papel Atribuído na Base</sup>',
                font=dict(size=14, family='Inter, sans-serif', color='#1e293b'),
                x=0.05
            ),
            showlegend=True,
            legend=dict(
                orientation="h",
                yanchor="bottom",
                y=1.02,
                xanchor="right",
                x=1
            ),
            hovermode='closest',
            margin=dict(b=20, l=20, r=20, t=80),
            xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
            yaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
            plot_bgcolor='#ffffff',
            paper_bgcolor='#ffffff'
        )
    )

    fig.show()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import MultiLabelBinarizer
```

```python
text_cols = ['trajetoria', 'expectativa', 'soft_skills', 'motivacao', 'areas_interesse']
df['texto_perfil'] = df[text_cols].fillna('').agg(' '.join, axis=1)

tfidf = TfidfVectorizer(max_features=100, stop_words=None)
tfidf_matrix = tfidf.fit_transform(df['texto_perfil'])

sim_matrix = cosine_similarity(tfidf_matrix, tfidf_matrix)
print("Shape da matriz de similaridade de texto:", sim_matrix.shape)
print("Exemplo de valores de similaridade:", sim_matrix[0, :5])
```

```python
df['disponibilidade_lista'] = df['disponibilidade'].fillna('').apply(
    lambda x: [item.strip() for item in str(x).split(',') if item.strip()]
)
mlb_disp = MultiLabelBinarizer()
disp_arr = mlb_disp.fit_transform(df['disponibilidade_lista'])
disp_cols = [f"disp_{c}" for c in mlb_disp.classes_]
disp_df = pd.DataFrame(disp_arr, columns=disp_cols, index=df.index)


text_cols = ['trajetoria', 'expectativa', 'soft_skills', 'motivacao', 'areas_interesse']
df['texto_perfil'] = df[text_cols].fillna('').agg(' '.join, axis=1)

tfidf = TfidfVectorizer(max_features=150, stop_words=None)
text_features = tfidf.fit_transform(df['texto_perfil']).toarray()

df_completo = pd.concat([df, disp_df, pd.DataFrame(text_features, index=df.index, columns=[f"txt_{i}" for i in range(text_features.shape[1])])], axis=1)
text_cols_names = [f"txt_{i}" for i in range(text_features.shape[1])]
```

```python
def calcular_compatibilidade_perfil(mentora, mentorada, disp_cols, text_cols_names):
    if mentora['instituicao'] != mentorada['instituicao'] or mentora['curso'] != mentorada['curso']:
        return 0.0
    if mentora['ano_ingresso'] >= mentorada['ano_ingresso']:
        return 0.0


    v_mentora_disp = mentora[disp_cols].values.astype(float)
    v_mentorada_disp = mentorada[disp_cols].values.astype(float)
    inter = np.sum(v_mentora_disp * v_mentorada_disp)
    uniao = np.sum(np.maximum(v_mentora_disp, v_mentorada_disp))
    sim_disp = inter / uniao if uniao > 0 else 0.0


    v_mentora_txt = mentora[text_cols_names].values.astype(float)
    v_mentorada_txt = mentorada[text_cols_names].values.astype(float)
    norm_m = np.linalg.norm(v_mentora_txt)
    norm_mo = np.linalg.norm(v_mentorada_txt)
    sim_txt = np.dot(v_mentora_txt, v_mentorada_txt) / (norm_m * norm_mo) if norm_m > 0 and norm_mo > 0 else 0.0

    score_final = (0.7 * sim_txt) + (0.3 * sim_disp)
    return score_final
```

```python
def analisar_similaridade_instituicao(df_full, instituicao_alvo, disp_cols, text_cols_names, threshold=0.3):
    df_inst = df_full[df_full['instituicao'].str.upper() == instituicao_alvo.upper()].copy()
    if len(df_inst) <= 1:
        return 0, 0, np.array([[]])

    mediana = df_inst['ano_ingresso'].median()
    mentoras = df_inst[df_inst['ano_ingresso'] <= mediana]
    mentoradas = df_inst[df_inst['ano_ingresso'] > mediana]

    if len(mentoras) == 0 or len(mentoradas) == 0:
        return 0, 0, np.array([[]])

    matriz = np.zeros((len(mentoras), len(mentoradas)))
    mentoras_l = list(mentoras.iterrows())
    mentoradas_l = list(mentoradas.iterrows())

    pares_similares = 0
    for i, (_, mentora) in enumerate(mentoras_l):
        for j, (_, mentorada) in enumerate(mentoradas_l):
            score = calcular_compatibilidade_perfil(mentora, mentorada, disp_cols, text_cols_names)
            matriz[i, j] = score
            if score >= threshold:
                pares_similares += 1

    return len(mentoras), len(mentoradas), pares_similares, matriz

m, md, pares, mat = analisar_similaridade_instituicao(df_completo, 'UPE', disp_cols, text_cols_names, threshold=0.25)
print(f"UPE -> Mentoras: {m}, Mentoradas: {md}, Pares com alta similaridade de perfil: {pares}")
```

```python
def gerar_rede_plotly_por_instituicao(df_full, instituicao_alvo, disp_cols, text_cols_names, threshold=0.25):
    df_inst = df_full[df_full['instituicao'].str.upper() == instituicao_alvo.upper()].copy()

    if len(df_inst) > 1:
        mediana_ingresso = df_inst['ano_ingresso'].median()
        mentoras = df_inst[df_inst['ano_ingresso'] <= mediana_ingresso]
        mentoradas = df_inst[df_inst['ano_ingresso'] > mediana_ingresso]
    else:
        mentoras = pd.DataFrame()
        mentoradas = pd.DataFrame()

    if len(mentoras) == 0 or len(mentoradas) == 0:
        print(f"[{instituicao_alvo}] Sem mentoras ou mentoradas suficientes.")
        return

    G = nx.Graph()
    for _, row in mentoras.iterrows():
        G.add_node(str(row['id']), papel='Mentora', curso=row['curso'])
    for _, row in mentoradas.iterrows():
        G.add_node(str(row['id']), papel='Mentorada', curso=row['curso'])

    mentoras_list = list(mentoras.iterrows())
    mentoradas_list = list(mentoradas.iterrows())

    pares_similares = 0
    for _, mentora in mentoras_list:
        for _, mentorada in mentoradas_list:
            if mentora['curso'] == mentorada['curso'] and mentora['ano_ingresso'] < mentorada['ano_ingresso']:
                v_mentora_disp = mentora[disp_cols].values.astype(float)
                v_mentorada_disp = mentorada[disp_cols].values.astype(float)
                inter = np.sum(v_mentora_disp * v_mentorada_disp)
                uniao = np.sum(np.maximum(v_mentora_disp, v_mentorada_disp))
                sim_disp = inter / uniao if uniao > 0 else 0.0

                v_mentora_txt = mentora[text_cols_names].values.astype(float)
                v_mentorada_txt = mentorada[text_cols_names].values.astype(float)
                norm_m = np.linalg.norm(v_mentora_txt)
                norm_mo = np.linalg.norm(v_mentorada_txt)
                sim_txt = np.dot(v_mentora_txt, v_mentorada_txt) / (norm_m * norm_mo) if norm_m > 0 and norm_mo > 0 else 0.0

                score = (0.7 * sim_txt) + (0.3 * sim_disp)
                if score >= threshold:
                    G.add_edge(str(mentora['id']), str(mentorada['id']), weight=score)
                    pares_similares += 1

    print(f"[{instituicao_alvo}] Mentoras: {len(mentoras)} | Mentoradas: {len(mentoradas)} | Pares similares: {pares_similares}")

    pos = nx.spring_layout(G, seed=42)

    edge_x = []
    edge_y = []
    for edge in G.edges():
        x0, y0 = pos[edge[0]]
        x1, y1 = pos[edge[1]]
        edge_x.extend([x0, x1, None])
        edge_y.extend([y0, y1, None])

    edge_trace = go.Scatter(x=edge_x, y=edge_y, line=dict(width=1.5, color='#888'), hoverinfo='none', mode='lines')

    node_x = []
    node_y = []
    node_color = []
    node_text = []
    for node in G.nodes():
        x, y = pos[node]
        node_x.append(x)
        node_y.append(y)
        papel = G.nodes[node]['papel']
        node_color.append('#87CEEB' if papel == 'Mentora' else '#90EE90')
        node_text.append(f"ID: {node}<br>Papel: {papel}<br>Curso: {G.nodes[node]['curso']}")

    node_trace = go.Scatter(
        x=node_x, y=node_y, mode='markers', hoverinfo='text',
        marker=dict(showscale=False, color=node_color, size=20, line=dict(width=2, color='#333'))
    )
    node_trace.text = node_text

    fig = go.Figure(data=[edge_trace, node_trace],
                    layout=go.Layout(
                        title=f'Rede Interativa de Similaridade - SWITAB ({instituicao_alvo})',
                        showlegend=False,
                        hovermode='closest',
                        margin=dict(b=20,l=5,r=5,t=40),
                        xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
                        yaxis=dict(showgrid=False, zeroline=False, showticklabels=False))
                    )
    fig.show()
```

## Abordagem Evolutiva (Algoritmo Genético / Cromossomos) para Otimização Combinatória

```python
def checa_mesma_instituicao(mentora, mentorada):
    return mentora['instituicao'].strip().upper() == mentorada['instituicao'].strip().upper()

def checa_mesmo_curso(mentora, mentorada):
    return mentora['curso'].strip().upper() == mentorada['curso'].strip().upper()

def checa_precedencia_periodo(mentora, mentorada):
    try:
        return float(mentora['ano_ingresso']) < float(mentorada['ano_ingresso'])
    except:
        return True

def construir_matriz_regras(mentoras_df, mentoradas_df, disp_cols, text_cols_names):
    m_list = list(mentoras_df.iterrows())
    md_list = list(mentoradas_df.iterrows())
    matriz = np.zeros((len(m_list), len(md_list)))

    for i, (_, mentora) in enumerate(m_list):
        for j, (_, mentorada) in enumerate(md_list):
            if (
                checa_mesma_instituicao(mentora, mentorada) and
                checa_mesmo_curso(mentora, mentorada) and
                checa_precedencia_periodo(mentora, mentorada)
            ):
                score = calcular_compatibilidade_perfil(mentora, mentorada, disp_cols, text_cols_names)
                matriz[i, j] = score
            else:
                matriz[i, j] = 0.0

    return m_list, md_list, matriz

def reparar_cromossomo(cromossomo, matriz_scores, capacidade_maxima):
    num_mentoras, num_mentoradas = matriz_scores.shape
    alocacoes = {}
    novo_cromossomo = cromossomo.copy()
    mentoradas_sem_vaga = []

    for j in range(num_mentoradas):
        i = novo_cromossomo[j]
        if matriz_scores[i, j] == 0.0:
            mentoradas_sem_vaga.append(j)
        else:
            alocacoes[i] = alocacoes.get(i, 0) + 1
            if alocacoes[i] > capacidade_maxima:
                alocacoes[i] -= 1
                mentoradas_sem_vaga.append(j)

    for j in mentoradas_sem_vaga:
        validas = np.where(matriz_scores[:, j] > 0)[0]
        disponiveis = [i for i in validas if alocacoes.get(i, 0) < capacidade_maxima]

        if disponiveis:
            melhor_i = disponiveis[np.argmax(matriz_scores[disponiveis, j])]
            novo_cromossomo[j] = melhor_i
            alocacoes[melhor_i] = alocacoes.get(melhor_i, 0) + 1
        else:
            if len(validas) > 0:
                melhor_i = validas[np.argmax(matriz_scores[validas, j])]
                novo_cromossomo[j] = melhor_i
                alocacoes[melhor_i] = alocacoes.get(melhor_i, 0) + 1
            else:
                novo_cromossomo[j] = np.random.randint(0, num_mentoras)

    return novo_cromossomo

def criar_cromossomo_inteligente(num_mentoradas, num_mentoras, matriz_scores, capacidade_maxima):
    cromossomo = np.zeros(num_mentoradas, dtype=int)
    for j in range(num_mentoradas):
        validas = np.where(matriz_scores[:, j] > 0)[0]
        if len(validas) > 0:
            cromossomo[j] = np.random.choice(validas)
        else:
            cromossomo[j] = np.random.randint(0, num_mentoras)
    return reparar_cromossomo(cromossomo, matriz_scores, capacidade_maxima)

def calcular_fitness(cromossomo, matriz_scores, capacidade_maxima):
    fitness_total = 0.0
    alocacoes = {}
    for idx_mentorada, idx_mentora in enumerate(cromossomo):
        score = matriz_scores[idx_mentora, idx_mentorada]
        if score == 0.0:
            fitness_total -= 0.5
            continue

        alocacoes[idx_mentora] = alocacoes.get(idx_mentora, 0) + 1
        if alocacoes[idx_mentora] > capacidade_maxima:
            fitness_total -= 0.2
            continue

        fitness_total += score
    return fitness_total
```

```python
def selecionar_pais(populacao, scores, k=3):
    selecionados = []
    for _ in range(2):
        inicios = np.random.choice(len(populacao), k, replace=False)
        melhor = inicios[0]
        for idx in inicios[1:]:
            if scores[idx] > scores[melhor]:
                melhor = idx
        selecionados.append(populacao[melhor])
    return selecionados[0], selecionados[1]

def crossover(pai1, pai2):
    ponto = np.random.randint(1, len(pai1))
    filho1 = np.concatenate([pai1[:ponto], pai2[ponto:]])
    filho2 = np.concatenate([pai2[:ponto], pai1[ponto:]])
    return filho1, filho2

def mutacao(cromossomo, num_mentoras, matriz_scores, capacidade_maxima, taxa_mutacao=0.15):
    filho = cromossomo.copy()
    for i in range(len(filho)):
        if np.random.rand() < taxa_mutacao:
            validas = np.where(matriz_scores[:, i] > 0)[0]
            if len(validas) > 0:
                filho[i] = np.random.choice(validas)
            else:
                filho[i] = np.random.randint(0, num_mentoras)
    return reparar_cromossomo(filho, matriz_scores, capacidade_maxima)
```

```python
def executar_ag_switab(matriz_scores, num_geracoes=150, tam_populacao=100, capacidade_maxima=2, taxa_mutacao=0.15):
    num_mentoras, num_mentoradas = matriz_scores.shape
    if num_mentoras == 0 or num_mentoradas == 0:
        return None, 0.0

    populacao = [criar_cromossomo_inteligente(num_mentoradas, num_mentoras, matriz_scores, capacidade_maxima) for _ in range(tam_populacao)]
    melhor_cromossomo = None
    melhor_fitness = -float('inf')

    for geracao in range(num_geracoes):
        scores = [calcular_fitness(ind, matriz_scores, capacidade_maxima) for ind in populacao]

        for idx, score in enumerate(scores):
            if score > melhor_fitness:
                melhor_fitness = score
                melhor_cromossomo = populacao[idx].copy()

        nova_populacao = []
        while len(nova_populacao) < tam_populacao:
            p1, p2 = selecionar_pais(populacao, scores)
            f1, f2 = crossover(p1, p2)
            f1 = mutacao(f1, num_mentoras, matriz_scores, capacidade_maxima, taxa_mutacao)
            f2 = mutacao(f2, num_mentoras, matriz_scores, capacidade_maxima, taxa_mutacao)
            nova_populacao.extend([f1, f2])

        populacao = nova_populacao[:tam_populacao]

    return melhor_cromossomo, melhor_fitness
```

```python
mediana = df_completo['ano_ingresso'].median()
mentoras_df = df_completo[df_completo['ano_ingresso'] <= mediana]
mentoradas_df = df_completo[df_completo['ano_ingresso'] > mediana]

m_list, md_list, matriz_final = construir_matriz_regras(mentoras_df, mentoradas_df, disp_cols, text_cols_names)
melhor_solucao, melhor_fit = executar_ag_switab(matriz_final, capacidade_maxima=2)

print(f"Melhor Fitness com Regras Reais (Corrigido): {melhor_fit:.4f}")
```

```python
def gerar_rede_plotly_por_instituicao(melhor_solucao, matriz_final, m_list, md_list, instituicao_alvo):
    G = nx.Graph()

    pares_incluidos = 0
    for idx_mentorada, idx_mentora in enumerate(melhor_solucao):
        _, mentorada_row = md_list[idx_mentorada]
        _, mentora_row = m_list[idx_mentora]

        # Filtra pela instituição desejada
        if mentora_row['instituicao'].strip().upper() != instituicao_alvo.strip().upper():
            continue

        score = matriz_final[idx_mentora, idx_mentorada]
        m_id = str(mentora_row['id'])
        md_id = str(mentorada_row['id'])

        if not G.has_node(m_id):
            G.add_node(m_id, papel='Mentora', curso=mentora_row['curso'], instituicao=mentora_row['instituicao'])
        if not G.has_node(md_id):
            G.add_node(md_id, papel='Mentorada', curso=mentorada_row['curso'], instituicao=mentorada_row['instituicao'])

        G.add_edge(m_id, md_id, weight=score)
        pares_incluidos += 1

    if len(G.nodes) == 0:
        print(f"Nenhum par encontrado para a instituição: {instituicao_alvo}")
        return

    print(f"[{instituicao_alvo}] Nós na rede: {len(G.nodes)} | Pares conectados: {pares_incluidos}")

    pos = nx.spring_layout(G, seed=42)

    edge_x = []
    edge_y = []
    for edge in G.edges():
        x0, y0 = pos[edge[0]]
        x1, y1 = pos[edge[1]]
        edge_x.extend([x0, x1, None])
        edge_y.extend([y0, y1, None])

    edge_trace = go.Scatter(
        x=edge_x, y=edge_y,
        line=dict(width=1.5, color='#888'),
        hoverinfo='none',
        mode='lines'
    )

    node_x = []
    node_y = []
    node_color = []
    node_text = []

    for node in G.nodes():
        x, y = pos[node]
        node_x.append(x)
        node_y.append(y)
        papel = G.nodes[node]['papel']
        node_color.append('#87CEEB' if papel == 'Mentora' else '#90EE90')
        node_text.append(f"ID: {node}<br>Papel: {papel}<br>Curso: {G.nodes[node]['curso']}<br>Inst: {G.nodes[node]['instituicao']}")

    node_trace = go.Scatter(
        x=node_x, y=node_y,
        mode='markers+text',
        text=[node for node in G.nodes()],
        textposition="top center",
        hoverinfo='text',
        marker=dict(
            showscale=False,
            color=node_color,
            size=22,
            line=dict(width=2, color='#333')
        )
    )
    node_trace.text = node_text

    fig = go.Figure(
        data=[edge_trace, node_trace],
        layout=go.Layout(
            title=f'Rede Otimizada de Mentoria - SWITAB ({instituicao_alvo})',
            showlegend=False,
            hovermode='closest',
            margin=dict(b=20, l=5, r=5, t=40),
            xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
            yaxis=dict(showgrid=False, zeroline=False, showticklabels=False)
        )
    )
    fig.show()
```

### Redes aleatorias

```python
def consolidar_metricas_redes(resultados_por_instituicao, stats_baseline):
    """
    Consolida as métricas topológicas, de compatibilidade e taxa de sucesso
    para cada rede institucional, comparando-as com o baseline.
    """
    dados_comparativos = []

    for inst, dados in resultados_por_instituicao.items():
        G = dados['grafo']
        m = dados['metricas']

        total_pares = m['total_pares']
        violacoes = m['violacoes_periodo']


        taxa_sucesso = ((total_pares - violacoes) / total_pares) * 100 if total_pares > 0 else 0.0


        graus = [d for n, d in G.degree()]
        grau_medio = np.mean(graus) if graus else 0.0
        grau_max = max(graus) if graus else 0


        maior_hub = max(G.nodes(), key=lambda n: G.degree(n)) if G.number_of_nodes() > 0 else "N/A"


        compat_baseline = stats_baseline['compatibilidade_media']
        compat_atual = m['compatibilidade_media']
        ganho_compatibilidade = ((compat_atual - compat_baseline) / compat_baseline) * 100 if compat_baseline > 0 else 0.0

        dados_comparativos.append({
            'Instituição': inst,
            'Total de Pares': total_pares,
            'Densidade': m['densidade'],
            'Grau Médio': grau_medio,
            'Grau Máximo (Hub)': grau_max,
            'ID do Principal Hub': maior_hub,
            'Compatibilidade Média': compat_atual,
            'Ganho vs Baseline (%)': ganho_compatibilidade,
            'Violações de Período': violacoes,
            'Taxa de Sucesso (%)': taxa_sucesso,
            'Coeficiente de Aglomeração': m['clustering_medio']
        })

    df_comparativo = pd.DataFrame(dados_comparativos)
    return df_comparativo


df_metricas_redes = consolidar_metricas_redes(resultados_por_instituicao, stats_baseline)
df_metricas_redes.style.format({
    'Densidade': '{:.4f}',
    'Grau Médio': '{:.2f}',
    'Compatibilidade Média': '{:.4f}',
    'Ganho vs Baseline (%)': '{:+.2f}%',
    'Taxa de Sucesso (%)': '{:.1f}%',
    'Coeficiente de Aglomeração': '{:.4f}'
})
```

### Distância de Similaridade

```python
def calcular_consolidado_metricas_rede(resultados_dict, threshold=0.25):
    """Consolida métricas de similaridade e estruturais de rede bipartida para o conjunto de instituições."""
    dados = []

    for inst, dados_inst in resultados_dict.items():

        lista_mentoras, lista_mentoradas, total_similares, matriz = dados_inst

        media_sim = float(np.mean(matriz)) if matriz.size > 0 else 0.0
        num_mentoras, num_mentoradas = matriz.shape


        G = nx.Graph()
        nodes_mentoras = [f"Mentora_{i}" for i in range(num_mentoras)]
        nodes_mentoradas = [f"Mentorada_{j}" for j in range(num_mentoradas)]

        G.add_nodes_from(nodes_mentoras, bipartite=0)
        G.add_nodes_from(nodes_mentoradas, bipartite=1)

        for i in range(num_mentoras):
            for j in range(num_mentoradas):
                score = matriz[i, j]
                if score >= threshold:
                    G.add_edge(nodes_mentoras[i], nodes_mentoradas[j], weight=score)


        densidade = nx.density(G) if G.number_of_nodes() > 0 else 0.0
        graus = [d for n, d in G.degree()]
        grau_medio = float(np.mean(graus)) if graus else 0.0
        grau_maximo = int(np.max(graus)) if graus else 0


        if graus:
            limiar_hub = np.mean(graus) + np.std(graus)
            hubs = [n for n, d in G.degree() if d > limiar_hub]
            qtd_hubs = len(hubs)
        else:
            qtd_hubs = 0

        dados.append({
            'Instituição': inst,
            'Total Pares Similares': total_similares,
            'Média de Similaridade': media_sim,
            'Densidade': densidade,
            'Grau Médio': grau_medio,
            'Grau Máximo (Hub)': grau_maximo,
            'Qtd Hubs': qtd_hubs,
            'Shape da Matriz': f"({num_mentoras}, {num_mentoradas})"
        })

    return pd.DataFrame(dados)

dicionario_resultados = {
    'UFRPE': distancia_similaridade_ufrpe,
    'UPE': distancia_similaridade_upe,
    'UNICAP': distancia_similaridade_unip,
    'Cesar School': distancia_similaridade_cesar
}

df_metricas_distancia_similaridade_consolidadas = calcular_consolidado_metricas_rede(dicionario_resultados, threshold=0.25)
df_metricas_distancia_similaridade_consolidadas
```

### Evolutivo

```python
def calcular_consolidado_evolutivo_automatico(df_completo, melhor_solucao_global, melhor_fit_global, disp_cols, text_cols_names, baseline_global=0.3361):
    dados = []
    mediana = df_completo['ano_ingresso'].median()

    mentoras_df_geral = df_completo[df_completo['ano_ingresso'] <= mediana]
    mentoradas_df_geral = df_completo[df_completo['ano_ingresso'] > mediana]

    m_list_geral, md_list_geral, matriz_final_geral = construir_matriz_regras(mentoras_df_geral, mentoradas_df_geral, disp_cols, text_cols_names)

    instituicoes = df_completo['instituicao'].dropna().unique()

    for inst in instituicoes:
        df_inst = df_completo[df_completo['instituicao'].str.strip().str.upper() == inst.strip().upper()].copy()
        if df_inst.empty:
            continue

        mentoras_df = df_inst[df_inst['ano_ingresso'] <= mediana]
        mentoradas_df = df_inst[df_inst['ano_ingresso'] > mediana]

        if mentoras_df.empty or mentoradas_df.empty:
            continue

        _, _, matriz_inst = construir_matriz_regras(mentoras_df, mentoradas_df, disp_cols, text_cols_names)
        num_mentoras, num_mentoradas = matriz_inst.shape

        G = nx.Graph()
        nodes_mentoras = [f"Mentora_{i}" for i in range(num_mentoras)]
        nodes_mentoradas = [f"Mentorada_{j}" for j in range(num_mentoradas)]

        G.add_nodes_from(nodes_mentoras, bipartite=0)
        G.add_nodes_from(nodes_mentoradas, bipartite=1)

        total_pares = 0
        compat_scores = []
        violacoes = 0

        for idx_mentorada_global, idx_mentora_global in enumerate(melhor_solucao_global):
            _, mentorada_row = md_list_geral[idx_mentorada_global]
            _, mentora_row = m_list_geral[idx_mentora_global]

            if mentora_row['instituicao'].strip().upper() == inst.strip().upper():
                total_pares += 1
                score = matriz_final_geral[idx_mentora_global, idx_mentorada_global]

                if score > 0.0:
                    try:
                        idx_mentora_local = list(mentoras_df.index).index(mentora_row.name)
                        idx_mentorada_local = list(mentoradas_df.index).index(mentorada_row.name)

                        node_id_mentora = f"Mentora_{idx_mentora_local}"
                        node_id_mentorada = f"Mentorada_{idx_mentorada_local}"

                        # Armazenando o ID real do dataframe original como atributo do nó
                        G.add_node(node_id_mentora, real_id=mentora_row.name)
                        G.add_node(node_id_mentorada, real_id=mentorada_row.name)

                        G.add_edge(node_id_mentora, node_id_mentorada, weight=score)
                    except:
                        pass
                    compat_scores.append(score)
                else:
                    violacoes += 1

        if total_pares == 0:
            continue

        compat_atual = float(np.mean(compat_scores)) if compat_scores else 0.0
        ganho_compatibilidade = ((compat_atual - baseline_global) / baseline_global) * 100 if baseline_global > 0 else 0.0
        taxa_sucesso = ((total_pares - violacoes) / total_pares) * 100 if total_pares > 0 else 0.0

        graus = [d for n, d in G.degree()]
        grau_medio = np.mean(graus) if graus else 0.0
        grau_max = max(graus) if graus else 0

        if G.number_of_nodes() > 0:
            maior_no = max(G.nodes(), key=lambda n: G.degree(n))
            maior_hub = G.nodes[maior_no].get('real_id', maior_no)
        else:
            maior_hub = "N/A"

        densidade = nx.density(G) if G.number_of_nodes() > 0 else 0.0
        clustering_medio = nx.average_clustering(G) if G.number_of_nodes() > 0 else 0.0

        dados.append({
            'Instituição': inst,
            'Total de Pares': total_pares,
            'Densidade': densidade,
            'Grau Médio': grau_medio,
            'Grau Máximo (Hub)': grau_max,
            'ID do Principal Hub': maior_hub,
            'Compatibilidade Média': compat_atual,
            'Ganho vs Baseline (%)': ganho_compatibilidade,
            'Violações de Período': violacoes,
            'Taxa de Sucesso (%)': taxa_sucesso,
            'Coeficiente de Aglomeração': clustering_medio
        })

    return pd.DataFrame(dados)

df_metricas_evolutivas_consolidadas = calcular_consolidado_evolutivo_automatico(
    df_completo, melhor_solucao, melhor_fit, disp_cols, text_cols_names
)
df_metricas_evolutivas_consolidadas
```
